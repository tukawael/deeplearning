```python
import numpy as np
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
 
def tanh(x):
    return np.tanh(x)
    
Wf = 0.7
Uf = 0.5
bf = 0.1
 
Wi = 0.6
Ui = 0.4
bi = 0.1
 
Wc = 0.5
Uc = 0.3
bc = 0.0
 
Wo = 0.6
Uo = 0.4
bo = 0.1
 
Wy = 4.0
by = 0.1
 
h = 0.0
c = 0.0

sequence = [1, 2, 3]
 
 
print("LSTM Numerical Example")
print("Input sequence:", sequence, " -> predict next value")


for t, x in enumerate(sequence, start=1):
    f = sigmoid(Wf * x  +  Uf * h  +  bf)
    i = sigmoid(Wi * x  +  Ui * h  +  bi)
    c_tilde = tanh(Wc * x  +  Uc * h  +  bc)
    c = f * c  +  i * c_tilde

    o = sigmoid(Wo * x  +  Uo * h  +  bo)
 
    h = o * tanh(c)
 
    print(f"\nTime step t = {t},  x = {x}")
    print(f"  forget gate   f  = {f:.4f}")
    print(f"  input gate    i  = {i:.4f}")
    print(f"  candidate     c~ = {c_tilde:.4f}")
    print(f"  cell state    c  = {c:.4f}")
    print(f"  output gate   o  = {o:.4f}")
    print(f"  hidden state  h  = {h:.4f}")

y_hat = Wy * h + by

print("Output layer:  y_hat = Wy * h + by")
print(f"             = {Wy} * {h:.4f} + {by}")
print(f"\nPredicted next value = {y_hat:.2f}")
print("Actual next value    = 4")
print("(model output ≈ 3.8, close to 4 -- pattern learned)")
```
